````python
import cirq


q = cirq.LineQubit(0) #LineQubit is used in Cirq , can be changed while using Qiskit as Qubit/qubit

                         # Creating a circuit with H and M (hadmard Superposition and Measurement)
circuit = cirq.Circuit(
    cirq.H(q),
    cirq.measure(q, key="result")
)

print(circuit)

# Run the simulator
simulator = cirq.Simulator()
result = simulator.run(circuit, repetitions=5000)

print(result)
print(result.histogram(key="result"))
````

Executes the circuit 5000 times. Since quantum measurements are probabilistic, running multiple repetitions lets you observe the distribution of outcomes.

````python
import math

N = 15
a = 2

print(f"Trying to factor N = {N}")
print(f"Choose a = {a}\n")

for r in range(1, 20):
    value = pow(a, r, N)
    print(f"r = {r:2d}  →  {a}^{r} mod {N} = {value}")

    if value == 1:
        break

print("\nFound order r =", r)

if r % 2 == 0:
    x = pow(a, r // 2, N)

    factor1 = math.gcd(x - 1, N)
    factor2 = math.gcd(x + 1, N)

    print("Potential factors:")
    print(factor1)
    print(factor2)
````

Grover's Algorithm for
Classical search O(N)

Quantum search O(√N)

````python
import cirq

q0, q1 = cirq.LineQubit.range(2)

circuit = cirq.Circuit()

# Put qubits into superposition
circuit.append([
    cirq.H(q0),
    cirq.H(q1)
])

# Oracle for |11>
circuit.append(
    cirq.CZ(q0, q1)
)

# Grover diffusion operator
circuit.append([
    cirq.H(q0),
    cirq.H(q1),
    cirq.X(q0),
    cirq.X(q1),
    cirq.CZ(q0, q1),
    cirq.X(q0),
    cirq.X(q1),
    cirq.H(q0),
    cirq.H(q1)
])

# Measure
circuit.append(
    cirq.measure(q0, q1, key="result")
)

print(circuit)

simulator = cirq.Simulator()

result = simulator.run(
    circuit,
    repetitions=1000
)

print(result.histogram(key="result"))
````

````python
import oqs

algorithm = "ML-KEM-768"

with oqs.KeyEncapsulation(algorithm) as alice:

    # Alice creates her key pair
    public_key = alice.generate_keypair()

    with oqs.KeyEncapsulation(algorithm) as bob:

        # Bob encapsulates a secret using Alice's public key
        ciphertext, bob_secret = bob.encap_secret(public_key)

    # Alice decapsulates the ciphertext
    alice_secret = alice.decap_secret(ciphertext)

print("Algorithm:", algorithm)
print("Ciphertext length:", len(ciphertext))
print("Alice secret length:", len(alice_secret))
print("Bob secret length:", len(bob_secret))

print("\nShared secrets match:")
print(alice_secret == bob_secret)
````

ML-DSA Digital Signature

````python
import oqs

message = b"Transfer INR 1000 to Alice"

algorithm = "ML-DSA-65"

with oqs.Signature(algorithm) as signer:

    public_key = signer.generate_keypair()

    signature = signer.sign(message)

print("Algorithm:", algorithm)
print("Message:", message)
print("Public key size:", len(public_key))
print("Signature size:", len(signature))
````

````python
with oqs.Signature(algorithm) as verifier:

    valid = verifier.verify(
        message,
        signature,
        public_key
    )

print("Signature valid:", valid)
````

````python
import oqs
import os

from cryptography.hazmat.primitives.kdf.hkdf import HKDF
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.ciphers.aead import AESGCM
````

````python
with oqs.KeyEncapsulation("ML-KEM-768") as alice:

    alice_public_key = alice.generate_keypair()

    with oqs.KeyEncapsulation("ML-KEM-768") as bob:

        ciphertext, bob_secret = bob.encap_secret(
            alice_public_key
        )

    alice_secret = alice.decap_secret(ciphertext)

print("KEM shared secret:", alice_secret == bob_secret)
````

````python
def derive_key(shared_secret):   #derive AES Key

    hkdf = HKDF(
        algorithm=hashes.SHA256(),
        length=32,
        salt=None,
        info=b"PQC Workshop"
    )

    return hkdf.derive(shared_secret)
````

````python
alice_aes_key = derive_key(alice_secret)
bob_aes_key = derive_key(bob_secret)

print("AES keys match:", alice_aes_key == bob_aes_key)
````

````python
message = b"Confidential banking transaction"  #encryption

nonce = os.urandom(12)

aes = AESGCM(alice_aes_key)

encrypted_message = aes.encrypt(
    nonce,
    message,
    None
)

print("Encrypted message:")
print(encrypted_message.hex())
````

````python
aes = AESGCM(bob_aes_key)   #decrypt

decrypted_message = aes.decrypt(
    nonce,
    encrypted_message,
    None
)

print("Decrypted:")
print(decrypted_message.decode())
````

````python
message = b"Confidential banking transaction"

with oqs.Signature("ML-DSA-65") as signer:

    signing_public_key = signer.generate_keypair()

    signature = signer.sign(message)

print("Signature created.")
print("Signature size:", len(signature))
````

````python
with oqs.Signature("ML-DSA-65") as verifier:

    valid = verifier.verify(
        message,
        signature,
        signing_public_key
    )

print("Signature valid:", valid)
````

````python
tampered_message = b"Confidential banking transaction - MODIFIED"  #tampered

with oqs.Signature("ML-DSA-65") as verifier:

    valid = verifier.verify(
        tampered_message,
        signature,
        signing_public_key
    )

print("Tampered message valid:", valid)
````
